## 1.2 JSON Key Names
Which of the following objects have a well-formed JSON key name?

6. `{'name': 1}`

#### **Solution**

### 2.1.1 Document A

```
<?xml version="1.0" encoding="utf-8"?>

<catalog>

```

#### **Solution**

Document A has the following problems:
1. Comments `<!-- -->` cannot include the characters `--`;
2. The quotes in XML must always be single quotes or double quotes, but not "Word-style" quotes (〝, 〞, \`, etc.);
3. Attribute `version` in `publish_date` is duplicated, this is forbidden;
4. The `lang` attribute should be quoted;
5. `<` must be escaped in text. Also it is suggested to use `&gt;` for the `>` symbol;
6. The `book` start tag does not correspond to the `Book` end tag;
7. The `catalog` tag is not closed correctly;

`<xml_parse>true</xml_parse>` is already well-formed and does not need fixing. The underscore is part of the element name: this element has no namespace prefix. 

XML reserves names beginning with `xml` (in any capitalization) for standardization, so it's better to use another name such as `parse`.

This naming recommendation does not make the original element malformed. We keep `xml_parse` below to show that correcting the document does not require renaming it. 

Here is the corrected document:


# 3. XML Document Structure

Below is an empty table describing where different XML constructs can appear within an XML document. Fill in each cell with **yes** or **no** to indicate whether the given construct (elements, attributes, text, comments) is allowed in that position.

|                | Top-Level | Between Element Tags | Inside Opening Element Tag |
|----------------|-----------|----------------------|----------------------------|
| **Elements**   | ?         | ?                    | ?                          |
| **Attributes** | ?         | ?                    | ?                          |
| **Text**       | ?         | ?                    | ?                          |
| **Comments**   | ?         | ?                    | ?                          |

#### **Solution**

|                | Top-Level | Between Element Tags | Inside Opening Element Tag |
|----------------|-----------|----------------------|----------------------------|
| **Elements**   | yes (once)| yes                  | no                         |
| **Attributes** | no        | no                   | yes                        |
| **Text**       | no        | yes                  | no                         |
| **Comments**   | yes       | yes                  | no                         |

### Explanations

1. **Elements**  
   - **Top-Level**: An XML document must have exactly **one** root element, hence "yes (once)".  
   - **Between Element Tags**: Child elements can appear within another element, so "yes".  
   - **Inside Opening Element Tag**: Not possible; elements cannot be nested in the start tag itself, so "no".

2. **Attributes**  
   - **Top-Level**: Standalone attributes (not inside an element) are not allowed, so "no".  
   - **Between Element Tags**: Attributes cannot exist outside of a start tag, so "no".  
   - **Inside Opening Element Tag**: This is where attributes belong (e.g., `<myElement attr="value">`), so "yes".

3. **Text**  
   - **Top-Level**: Outside the single root element, only whitespace may be allowed, but not text content in general, so "no".  
   - **Between Element Tags**: Text content typically appears within elements, so "yes".  
   - **Inside Opening Element Tag**: Not allowed; only attributes can be placed there, so "no".

4. **Comments**  
   - **Top-Level**: Comments can be placed before or after the root element, or even outside of it, so "yes".  
   - **Between Element Tags**: Comments are allowed between elements (e.g., `<!-- comment -->`), so "yes".  
   - **Inside Opening Element Tag**: Comments cannot appear in the middle of a start tag, so "no".

## 5.2 Paths and Namespace Prefixes

The rest of this section uses `workspace/documents/catalog.xml`. It extends the Section 4 document with two more books, and its `publisher` element declares its own default namespace:

```xml
<catalog xmlns="http://example.org/catalog"
    xmlns:bk="http://example.org/book"

    <book id="b1" pub:year="1949">

    </publisher>

```

A path selects nodes step by step:

- `/` selects children, and `//` selects descendants at any depth.
- `@` selects attributes.
- `string()` at the end returns the text value of each selected node.

A step such as `cat:book` matches elements by **namespace URI and local name**. The query declares the prefix `cat` with `declare namespace`, and the prefix does not need to match the one in the document. Below, the document writes `bk:title`, but the query uses the prefix `book` for the same URI:

````python
%%jsoniq
declare namespace cat = "http://example.org/catalog";
declare namespace book = "http://example.org/book";

let $doc := fn:doc("workspace/documents/catalog.xml")
return $doc/cat:catalog/cat:book/book:title/string()
````

## 5.6 From XML to JSON

XML and JSON can carry the same data, but they are built from different pieces: XML has elements, attributes, and text, while JSON has objects, arrays, and values. Converting one into the other means deciding how each XML construct maps to JSON.

A JSONiq query can build JSON directly: `{ "key": value }` builds an object, and `[ ... ]` builds an array. The values can be path expressions. The query below converts the magazine in `catalog.xml`. In VS Code, use `workspace/xml-to-json.jq`. The query makes three decisions:

- Each `bk:magazine` element becomes one object, and the `for` inside `[ ]` collects these objects into an array.
- The attribute `issue` and the child elements `title` and `editor` all become keys, because JSON has no attributes.
- `/string()` extracts the text value of a node. XML text has no types, so `xs:integer()` converts the issue number; without it, the value would be the string `"42"`.

```json
  { "id": "b2", "title": "The Hobbit", "author": "J. R. R. Tolkien", "year": 1937 },
  { "id": "b3", "title": "To Kill a Mockingbird", "author": "Harper Lee", "year": 1960 }

```

When your query works, predict the result of each change below, then try it:

#### **Solution**

The books part follows the magazine pattern: `id` and `year` are attributes, `title` and `author` are child elements, and `xs:integer()` converts the year. Each step needs the right namespace: `title` is in the book namespace, `author` in the catalog namespace, and the attribute `year` in the publisher namespace.

````python
%%jsoniq
declare namespace cat = "http://example.org/catalog";
declare namespace bk = "http://example.org/book";
declare namespace pub = "http://example.org/publisher";

let $doc := fn:doc("workspace/documents/catalog.xml")
return {
  "magazines": [
    for $magazine in $doc/cat:catalog/bk:magazine
    return {
      "title": $magazine/cat:title/string(),
      "editor": $magazine/pub:editor/string(),
      "issue": xs:integer($magazine/@issue)
    }
  ],
  "books": [
    for $book in $doc/cat:catalog/cat:book
    return {
      "id": $book/@id/string(),
      "title": $book/bk:title/string(),
      "author": $book/cat:author/string(),
      "year": xs:integer($book/@pub:year)
    }
  ]
}
````

1. The value is the `bk:title` element itself, not its text. Jupyter shows it as the string `"<bk:title>1984</bk:title>"`, and VS Code shows the XML element, so the XML markup leaks into the JSON result. A path selects nodes, and `/string()` turns each node into its text.
2. Every book title becomes `null`. The book's `title` is in the book namespace, so `cat:title` selects nothing. Selecting nothing is not an error, and JSONiq writes an empty value in an object as `null`. A namespace mistake can therefore lose data without any error message.
