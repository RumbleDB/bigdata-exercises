## Exercise 2: Safe and Idempotent HTTP methods

HTTP methods are crucial in RESTful APIs and typically correspond to the main CRUD operations:
  - **GET** is used to *retrieve* a resource (Read).
  - **POST** is used to *create* a new resource (Create).
  - **PUT** is used to *update* an existing resource (Update).
  - **DELETE** is used to *remove* a resource (Delete).

<br>

- **Safe Methods**: An HTTP method is considered *safe* if the action it performs is read-only and does not modify the resource on the server. Making a "safe" request should not cause any side effects on the server data.

- **Idempotent Methods**: An HTTP method is *idempotent* if multiple **identical** requests in a row produce the same effect on the server as a single request would. In other words, sending the same request once or multiple times should give the same end result on the server state.

Complete the table below by indicating with **Yes** or **No** whether each method is **Idempotent** and/or **Safe**

| HTTP Method | Idempotent | Safe |
|-------------|-------------|-------|
| GET         | ?           | ?     |
| POST        | ?           | ?     |
| PUT         | ?           | ?     |
| DELETE      | ?           | ?     |

### Solution

| HTTP Method | Idempotent? | Safe? |
|-------------|-------------|-------|
| **GET**     | Yes         | Yes   |
| **POST**    | No          | No    |
| **PUT**     | Yes         | No    |
| **DELETE**  | Yes         | No    |

#### Explanations

1. **GET**  
   - **Safe**: A `GET` request is intended only to *retrieve* a resource without making any changes to it. Therefore, it does not modify the server state.  
   - **Idempotent**: Multiple `GET` requests to the same URL result in the same action (retrieving the resource) without changing its state. Hence, doing it once or several times yields the same effect on the server.

2. **POST**  
   - **Not Safe**: A `POST` request is often used to *create* a new resource by submitting data that changes the server state in some way (e.g., inserting a new record in a database).  
   - **Not Idempotent**: If you send the same `POST` request multiple times, it would create multiple separate resources.

3. **PUT**  
   - **Idempotent**: A `PUT` request usually *replaces* or *updates* a resource at a given URI. Sending the same `PUT` request multiple times with the same payload leads to the same final state.  
   - **Not Safe**: Even though it’s idempotent, a `PUT` modifies a resource, so it is not considered “safe” under HTTP definitions.

4. **DELETE**  
   - **Idempotent**: A `DELETE` request, if repeated multiple times, will remove the same resource. Subsequent `DELETE` requests do not further change the server’s state.  
   - **Not Safe**: Like `PUT`, it modifies the server’s state by removing a resource, so it is not a “read-only” operation.

For further information and explanations you can visit [Idempotent HTTP Methods](https://developer.mozilla.org/en-US/docs/Glossary/Idempotent) and [Safe HTTP Methods](https://developer.mozilla.org/en-US/docs/Glossary/Safe/HTTP)

## Exercise 3: Service Level Agreements

Cloud services usually come with a **Service Level Agreement (SLA)**: a contract describing the service quality the provider commits to delivering.

The main parameters are:

- **Availability:** The fraction of time the service can answer requests.
- **Durability:** The fraction of stored objects expected to survive without permanent loss over a year.
- **Latency:** The time it takes to respond to a request. It is often expressed as a percentile, such as “99.9% of requests are answered within 300 ms.”
- **Throughput:** The number of operations the service can handle per second.

For this exercise, we consider what Amazon S3 offers:

- Availability: $99.99\%$.
- Durability: $99.999999999\%$, also called “eleven nines.”
- Throughput: $5{,}500$ reads or $3{,}500$ writes per second.
- Typical latency: $100$ ms or more, with no guaranteed latency bound.

Recall the Sloan Digital Sky Survey (SDSS) dataset described in the textbook . It contains a total of **273 TB in 176,000,000 files**.

Throughout this exercise, use:

- Decimal units: $1\text{ TB} = 10^{12}$ bytes and $1\text{ Gb/s} = 10^9$ bits per second.
- A year of $365$ days.

$$

Each additional nine reduces the allowed downtime by a factor of ten.

$$

$$

The application therefore has approximately **$99.89\%$ availability**, corresponding to about **$9.6$ hours of downtime per year**.

**c.** The percentage alone does not specify a maximum outage length: we must also know the measurement period and how downtime is counted.

If $99.99\%$ availability is measured over one year, the total allowed downtime is about $52.6$ minutes. A single outage entirely within that year cannot exceed this without breaking the target.

However, the target does not say how downtime is distributed. It could be:

- One outage lasting about $52.6$ minutes.
- Many shorter outages with the same total duration.

An outage crossing the boundary between two measurement periods may count partly toward each period. It's also good to keep in mind that an SLA is a contractual commitment, not a guarantee that a longer outage can never occur.

### Task 2: Durability

**a.** You store all $176{,}000{,}000$ SDSS files on S3. How many objects would you expect to lose per year? Assuming a constant loss rate, approximately how long would you expect to wait for the first loss?

**b.** How many objects would you need to store for the expected loss to be one object per year?

**c.** Why can a provider offer durability with many more nines than availability?

#### Solution

**a.** With eleven nines of durability, the annual probability of losing an object is:

$$
1 - 0.99999999999 = 10^{-11}
$$

For the entire dataset, the expected number of lost objects per year is:

$$
176{,}000{,}000 \times 10^{-11} = 0.00176
$$

Under a constant-rate loss model, the approximate expected waiting time is:

$$
\frac{1}{0.00176} \approx 568\text{ years}
$$

That is roughly **one loss every $570$ years on average**.

**b.** We need enough objects for the expected annual loss to equal one:

$$
\text{Number of objects} \times 10^{-11} = 1
$$

Therefore:

$$
\text{Number of objects} = 10^{11}
$$

That is **100 billion objects**.

**c.** Durability concerns **permanent loss** while availability concerns **whether the data can be accessed right now**.

A provider can protect durability by keeping copies in multiple independent locations. An object is permanently lost only if all recoverable copies are lost.

The service can still be temporarily unavailable even when those copies are safe. For example:

- A network failure may prevent access.
- A software failure may stop requests from being processed.
- Heavy load may overwhelm the service.

Keeping more copies protects against data loss, but it does not eliminate every cause of temporary unavailability.

\text{Total time}

| Object size | Transfer time at $1\text{ Gb/s}$ | Total time, including $100$ ms latency | Main contribution |

**c.** Counting only the $100$ ms latency per request:

$$

$$
\frac{273 \times 10^{12} \times 8}{10^9}

$$
\text{Requests in progress}

Using only the $100$ ms latency:

$$
5{,}500 \times 0.1 = 550
$$

So approximately **$550$ requests** must be in progress at once.

$$
5{,}500 \times 1.55\text{ MB/s}
\approx 8.5\text{ GB/s}

$$
\frac{10^9}{1.55 \times 10^6 \times 8}

$$

This is an ideal bandwidth limit, ignoring overhead and assuming enough parallel requests to hide latency.

A single $1\text{ Gb/s}$ connection therefore cannot use the full operation rate. Reaching it requires enough aggregate bandwidth, for example through multiple machines or a much faster connection.

**e.** At most $0.1\%$ of requests may exceed $300$ ms:

$$
1{,}000{,}000 \times 0.001 = 1{,}000
$$

For a page requesting $100$ independent objects, the probability that all are fast is:

$$
0.999^{100} \approx 0.9048
$$

The probability of at least one slow object is therefore:

$$
1 - 0.999^{100} \approx 9.5\%
$$

With $1{,}000$ objects:

$$
1 - 0.999^{1000} \approx 63.2\%
$$

Even a small probability of a slow individual request can mean that many page loads wait for a slow object.

**f.** The response time seen by a client depends on factors such as:

- The distance between the client and the service.
- Network congestion and the client’s connection.
- The size of the object being transferred.
- Processing delays within the service.

The provider has more control over its own capacity to process operations than over the full path to every client.

For large objects, data transfer time may also dominate the initial request latency. Throughput and latency therefore describe different aspects of performance: a high operation rate does not mean that each request finishes quickly.

## Exercise 4: Key-Value Vector Clocks

> Reference: "Dynamo: Amazon’s Highly Available Key-value Store". In SOSP ’07 (Vol. 41, p. 205). [DOI](https://dl.acm.org/citation.cfm?doid=1294261.1294281)

Multiple distributed hash tables use vector clocks for capturing causality among different versions of the same object. In Amazon's Dynamo, a vector clock is associated with every version of every object.

Let $VC$ be an $N$-element array which contains non-negative integers, initialized to 0, representing $N$ logical clocks of the $N$ processes (nodes) of the system. $VC$ gets its $j$ element incremented by one everytime node $j$ performs a write operation on it. <br>
Moreover, $VC(x)$ denotes the vector clock of a write event, and $VC(x)_z$ denotes the element of that clock for the node $z$.

The formal definition of partial ordering that we get from using vector clocks is the following:
$$VC(x) \leq VC(y) \iff \forall z[VC(x)_z \leq VC(y)_z]$$

1. **Syntactic reconciliation**: If $VC(x) \leq VC(y)$ and $VC(x) \neq VC(y)$, then $x$ is an ancestor of $y$. That means that $y$ supersedes $x$, and the system can safely discard $x$ without losing information. If neither $VC(x) \leq VC(y)$ nor $VC(y) \leq VC(x)$ holds, then $x$ and $y$ are *concurrent* (causally unrelated), meaning they were written on parallel branches, and Dynamo keeps both.

2. **Semantic reconciliation.** A `get()` can return several concurrent versions, together with a *context*: the element-wise maximum of their vector clocks. The client merges the versions into one value (*reconciliation by the client*), for example by combining the items of two shopping carts, and passes the context back with its `put()`. The node that coordinates the write computes the clock of the new version in two steps:

    1. Take the element-wise maximum of the clocks of the versions the write is based on, i.e., the context. In a version DAG, these are the versions with an arrow into the new version.
    2. Increment the coordinating node's own entry by one.

The new version supersedes all the versions it is based on. For example, in Figure 3 of the Dynamo paper (with $S_x, S_y, S_z$ as nodes 0, 1, 2), a client reads the concurrent versions D3 $[2,1,0]$ and D4 $[2,0,1]$, merges them, and $S_x$ coordinates the write. The element-wise maximum is $[2,1,1]$, and incrementing entry 0 gives D5 $[3,1,1]$.

## Exercise 5: Merkle Trees

A hash tree or Merkle tree is a binary tree in which every leaf node gets as its label a data block and every non-leaf node is labelled with the cryptographic hash of the labels of its child nodes. 

Some Key-Value stores use Merkle trees for efficiently detecting inconsistencies in data between replicas. 

This works by exchanging first the root hash and comparing it with their own. If the hashes match, the replicas are synchronized. If they do not match, then the children of the node (in the Merkle tree) will be retrieved, and their hashes will be compared. This process continues until the inconsistent leaf(s) are identified.

Each of the following tasks shows a pair of Merkle trees for some object. For each pair, specify the process for identifying the data blocks that do not match or explain why the trees are not valid.

## Exercise 6: Consistent Hashing

In a distributed hash table such as Chord or Dynamo, both nodes and keys are assigned positions on a ring using a hash function.

In this exercise, IDs have **8 bits**, so their values range from $0$ to $255$. Positions wrap around modulo $256$: after $255$, we return to $0$.

Use the following rules:

- **Key placement:** A key belongs to the first node reached by moving **clockwise** from the key’s position. If a node is exactly at that position, the key belongs to that node. A node at position $n$ therefore owns the interval $(p, n]$, where $p$ is the previous node’s position. This interval is called the node’s **domain**. It excludes $p$ and includes $n$.
- **Replication:** With replication factor $N$, a key is stored on the first $N$ distinct nodes clockwise from its position. This ordered list is the key’s **preference list**. Its first node is the **coordinator**. Each node therefore stores its own domain and the $N-1$ domains immediately before it.
- **Requests:** A load balancer sends a request to a random node. If that node is not on the key’s preference list, it forwards the request to the coordinator. The coordinator contacts the other $N-1$ storage nodes and answers once it has $R$ responses for a read (`get`) or $W$ confirmations for a write (`put`). Its own copy counts toward this total.
- **Unreachable nodes:** The preference list contains only healthy, reachable nodes. An unreachable node is skipped and replaced by the next available node clockwise.

### Task 2: Replication and preference lists

Now set the replication factor to $N = 3$.

**a.** Give the preference list and coordinator for each key.

**b.** The load balancer sends `get(k2)` to A. With $R = 2$, describe what happens until the client receives an answer.

**c.** Which of the six keys does B store, either as coordinator or as another replica?

#### Solution

**a.**

| Key | $k_1$ | $k_2$ | $k_3$ | $k_4$ | $k_5$ | $k_6$ |
|--:|--:|--:|--:|--:|--:|--:|
| Preference list | A, B, C | C, D, E | E, A, B | A, B, C | C, D, E | D, E, A |
| Coordinator | A | C | E | A | C | D |

**b.**

1. A calculates the preference list for $k_2$: **C, D, E**.
2. A is not on the list, so it forwards the request to C.
3. C reads its own copy and requests copies from D and E.
4. Once C has its own copy and one response from D or E, it has the required $R = 2$ responses.
5. C compares the versions using their vector clocks.
6. C returns the version or versions that have not been superseded, together with the version context.

If the versions conflict, C may return more than one version so the client can resolve the conflict.

If C were unreachable, the request would go to the next reachable node, D, and the preference list would be adjusted.

**c.** B stores keys from three domains:

|--|--|--|

### Task 3: Nodes joining and leaving

Each part starts with the original five nodes, A, B, C, D, and E. Assume $N = 3$ unless stated otherwise.

**a.** F joins at position $100$.

- Which domain is split, and what are the new domains?
- Which data must F receive, and where can it come from?
- What is the new preference list for $k_2$? Which node no longer needs its copy?
- Which of the six keys does F store after the change?

**b.** D leaves gracefully, announcing its departure before disconnecting.

- With $N = 1$, which node takes over D’s domain, and what data must be transferred?
- With $N = 3$, does that node need a new copy of the keys in D’s domain?
- Which nodes need additional data to restore three copies of every key?
- What are the new preference lists for $k_5$, $k_6$, and $k_3$?

**c.** D crashes without warning.

- With $N = 1$, which keys are lost if D’s stored data cannot be recovered?
- With $N = 3$, where do copies of the keys in D’s domain survive? Which nodes need additional data?
- Why is this harder to handle than a graceful departure?

#### Solution

**a. F joins at position $100$.**

C’s domain, $(75, 130]$, is split into:

- **F’s domain:** $(75, 100]$, containing $25$ IDs.
- **C’s remaining domain:** $(100, 130]$, containing $30$ IDs.

F must store its own domain and the two preceding domains:

| Data F needs | F’s role | Possible source | Example keys |
|--|--|--|--|
| $(75, 100]$ | First copy: coordinator | C | $k_2$ |
| B’s domain, $(20, 75]$ | Second copy | B or another existing replica | None |
| A’s domain, $(240, 20]$ | Third copy | A, B, or C | $k_1$, $k_4$ |

The new preference list for $k_2$ is **F, C, D**. E is no longer on the list and can remove its copy once the transfer is complete.

F ends up storing **$k_1$, $k_2$, and $k_4$**.

The preference list for $k_1$ and $k_4$ changes from **A, B, C** to **A, B, F**, so C can also remove those copies once F has them.

**b. D leaves gracefully.**

With **$N = 1$**, E takes over D’s domain, $(130, 190]$. E’s enlarged domain becomes $(130, 240]$, and D transfers $k_6$ to E.

With **$N = 3$**, E already stores a copy of every key in D’s domain. It therefore needs no additional transfer to take over as coordinator for those keys.

However, D also stored replicas of the two preceding domains. To replace all copies that disappear when D leaves, these transfers are needed:

| Domain previously stored by D | Node receiving a replacement copy | Example keys |
|--|--|--|
| $(130, 190]$ | B | $k_6$ |
| $(75, 130]$ | A | $k_2$, $k_5$ |
| $(20, 75]$ | E | None |

The new preference lists are:

- $k_5$: **C, E, A**.
- $k_6$: **E, A, B**.
- $k_3$: **E, A, B**, unchanged.

Because D announces its departure, these transfers can finish before it disconnects.

**c. D crashes without warning.**

With **$N = 1$**, there is no other copy of the keys in D’s domain, $(130, 190]$. If D’s data cannot be recovered, those keys are lost. Among the example keys, this means **$k_6$**.

With **$N = 3$**, the keys in D’s own domain still have copies on **E and A**.

The system must restore the same replacement copies as in part b:

| Affected domain | Surviving copies | Node receiving a replacement copy | Example keys |
|--|--|--|--|
| $(130, 190]$ | E and A | B | $k_6$ |
| $(75, 130]$ | C and E | A | $k_2$, $k_5$ |
| $(20, 75]$ | B and C | E | None |

This produces the following preference lists:

- $k_6$: **E, A, B**.
- $k_2$ and $k_5$: **C, E, A**.
- Keys in $(20, 75]$: **B, C, E**.

A crash is harder to handle because:

- The system must detect that D is unreachable.
- Replacement copies must come from surviving nodes.
- Until repairs finish, affected keys have only two copies instead of three.

One more node failure could reduce an affected key to a single surviving copy. Losing both remaining copies before repair would lose the key.

With a graceful departure, D can stay online until replacement copies are ready.

### Task 4: Finding a key with finger tables

A finger table helps a node find a key without knowing the positions of every node.For example, the standard Chord protocol chooses fingers using powers-of-two distances in the ID space.

For this task, use a **simplified finger-table rule**: each node records the $1$st, $2$nd, $4$th, $8$th, and so on nodes clockwise from itself. Use powers of two smaller than the number of nodes.

When a node receives a lookup for position $p$, it follows these steps:

1. If $p$ is in its own domain, it identifies itself as the responsible node.
2. If $p$ is in its immediate successor’s domain, it identifies that successor as the responsible node.
3. Otherwise, it forwards the lookup to the farthest finger that is still before $p$ when moving clockwise.

Each forwarding in step 3 counts as one **hop**. Identifying the successor in step 2 does not add a hop in this exercise.

**a.** Give the finger table for each of the five nodes.

**b.** Trace these lookups and count their hops:

- $k_3$, starting at A.
- $k_1$, starting at C.

**c.** For this part, assume a larger ID space that can hold $1{,}024$ distinct nodes. Give a logarithmic upper bound on the number of lookup hops under the simplified rule.

Why does Dynamo not need finger tables? What trade-off does it make?

#### Solution

**a.**

| Node | 1st node clockwise | 2nd node clockwise | 4th node clockwise |
|--|--|--|--|
| A | B | C | E |
| B | C | D | A |
| C | D | E | B |
| D | E | A | C |
| E | A | B | D |

**b. Lookup of $k_3$ at position $200$, starting at A:**

1. Position $200$ is not in A’s domain or B’s domain.
2. A’s fingers are B at $75$, C at $130$, and E at $240$. C is the farthest finger before $200$, so A forwards to C.
3. At C, position $200$ is not in C’s domain or its successor D’s domain.
4. C forwards to D at $190$, its farthest finger before $200$.
5. D sees that $200$ belongs to its successor E and identifies E as the answer.

The lookup is forwarded **A → C → D**, and D identifies E.

**Total: two hops.**

**Lookup of $k_1$ at position $10$, starting at C:**

1. Moving clockwise from C, we pass D at $190$, E at $240$, wrap around through $0$, and reach $10`.
2. Of C’s fingers, E is the farthest one before $10$. B at $75$ lies beyond the target after wrapping around.
3. C forwards to E.
4. E sees that $10$ belongs to its successor A and identifies A as the answer.

The lookup is forwarded **C → E**, and E identifies A.

**Total: one hop.**

**c.** Under the simplified rule, each forwarding removes at least half the remaining node steps toward the target’s predecessor. A logarithmic upper bound for $1{,}024$ nodes is therefore:

$$
\log_2 1024 = 10\text{ hops}
$$

Dynamo takes a different approach: each node keeps a complete membership list, maintained through gossip. It can calculate a key’s preference list locally and forward the request directly to the coordinator.

The trade-off is:

- **Finger tables:** Less membership information per node, but a lookup may pass through several nodes.
- **Full membership lists:** Direct routing to the coordinator, but every node must store and maintain information about the entire cluster.

Full membership lists are practical for managed clusters of moderate size. Finger tables are better suited to very large peer-to-peer networks, where keeping a complete list at every node would be costly.

## Exercise 8: Quorums

Dynamo normally stores each key on **N nodes**, identified by the key’s *preference list*.

Two settings control when reads and writes can succeed:

* **R:** The minimum number of responses the coordinator must receive and compare before returning the result of a read (`get`).
* **W:** The minimum number of nodes that must confirm a write (`put`) or deletion (`delete`) before the coordinator reports success.

The coordinator’s own stored copy counts toward both R and W.

When **R + W > N**, the read and write sets must overlap, provided both operations use the **same set of N nodes**. This is a **quorum** configuration.

For example, with **N = 3, R = 2, and W = 2**, any two-node read set shares at least one node with any two-node write set. A read therefore contacts at least one node that acknowledged a completed write. This overlap helps the system discover updated data, although overlap alone does not guarantee strong consistency.

When some nodes are unreachable, Dynamo can use other reachable nodes farther around the ring as temporary replacements. These replacement nodes may not already have a copy of the key: they can return an empty response to a read or temporarily store a new write.

This approach is called a **sloppy quorum**. It allows requests to succeed when enough replacement nodes are reachable. However, reads and writes may use different sets of nodes, so **R + W > N no longer guarantees that their sets overlap**.


### Task 1: The quorum condition

In this exercise, each key is stored on five nodes, so the replication factor is $N = 5$.

A write (put) needs confirmation from $W$ nodes to succeed. A read (get) needs responses from $R$ nodes to succeed.

For this task, reads and writes use the same five storage nodes. Unreachable nodes are not replaced.

**a.** Complete the table below.

"Overlap guaranteed" means that every successful read must include at least one node that acknowledged a successful write.

| W | R | R + W > N? | Overlap guaranteed? | Unreachable nodes tolerated by a put | … by a get |
|--:|--:|--:|--:|--:|--:|
| 3 | 3 | | | | |
| 5 | 1 | | | | |
| 1 | 5 | | | | |
| 2 | 2 | | | | |
| 4 | 2 | | | | |
| 1 | 1 | | | | |

**b.** Explain why $R + W > N$ guarantees that the read and write sets overlap.

For $W = 2$ and $R = 2$, give an example of two nodes that acknowledge a write and two nodes that respond to a read, with no nodes shared between the two sets.

**c.** Which configuration would you choose for:

* A key that is read frequently but rarely updated?
* A key that is updated frequently but rarely read?

What trade-off does each choice involve?

**d.** For any replication factor $N$ and write requirement $W$:

* What is the smallest value of $R$ that satisfies the quorum condition?
* How many unreachable nodes can a write tolerate?
* How many unreachable nodes can a read tolerate?

#### Solution

**a.**

| $W$ | $R$ | Is $R + W > N$? | Is overlap guaranteed? | Unreachable nodes tolerated by a write ($N - W$) | Unreachable nodes tolerated by a read ($N - R$) |
| --: | --: | :-------------: | :--------------------: | -----------------------------------------------: | ----------------------------------------------: |
|   3 |   3 |  Yes ($6 > 5$)  |           Yes          |                                                2 |                                               2 |
|   5 |   1 |  Yes ($6 > 5$)  |           Yes          |                                                0 |                                               4 |
|   1 |   5 |  Yes ($6 > 5$)  |           Yes          |                                                4 |                                               0 |
|   2 |   2 |   No ($4 < 5$)  |           No           |                                                3 |                                               3 |
|   4 |   2 |  Yes ($6 > 5$)  |           Yes          |                                                1 |                                               3 |
|   1 |   1 |   No ($2 < 5$)  |           No           |                                                4 |                                               4 |

**b.** A successful write is stored on at least $W$ nodes. A successful read checks at least $R$ nodes.

If these two sets had no nodes in common, they would contain at least $R + W$ different nodes. When $R + W > N$, this is impossible: there are only $N$ storage nodes. The sets must therefore overlap.

With $W = 2$ and $R = 2$, however, the sets can be separate. For example:

* Nodes ${n_1, n_2}$ acknowledge the write.
* Nodes ${n_3, n_4}$ respond to the read.

If the new value is stored only on $n_1$ and $n_2$, this read will miss it.

**c.** For a key that is **read frequently but rarely updated**, choose $W = 5$ and $R = 1$:

* A read needs a response from only one node, so it can succeed even when four nodes are unreachable.
* A write needs confirmation from all five nodes, so a single unreachable node prevents writes from succeeding.

For a key that is **updated frequently but rarely read**, choose $W = 1$ and $R = 5$:

* A write succeeds as soon as one node confirms that it stored the value. It can therefore tolerate four unreachable nodes.
* A read needs responses from all five nodes, so a single unreachable node prevents reads from succeeding.

In each case, making one operation easier to complete makes the other more demanding.

**d.** The quorum condition is $R + W > N$. For a given $W$, the smallest integer value of $R$ that satisfies it is:

$$
R = N - W + 1
$$

The number of unreachable nodes each operation can tolerate is:

* **Writes:** $N - W$.
* **Reads:** $N - R$.

For $N = 5$, allowing both operations to tolerate two unreachable nodes requires:

* $W \le 3$, so a write can succeed with three reachable nodes.
* $R \le 3$, so a read can succeed with three reachable nodes.

The quorum condition also requires $R + W \ge 6$. The only choice satisfying all three conditions is:

$$
W = R = 3
$$

Both reads and writes therefore require a **majority**: three of the five nodes.

### Task 2: Sloppy quorums

A cluster has five nodes arranged clockwise in a ring: **A → B → C → D → E → A**.

The system uses these settings:

* **N = 3:** Each key normally has copies on three nodes.
* **R = 2:** A read needs responses from at least two nodes to succeed.
* **W = 2:** A write needs confirmation from at least two nodes to succeed.

The key **k** maps to a position between B and C. Its three usual storage nodes, called its *preference list*, are therefore **C, D, and E**.

Any request for k is forwarded to the first reachable node in that preference list. This node coordinates the request. If a storage node cannot be reached, the system skips it and uses the next available node clockwise around the ring. This replacement rule is called a **sloppy quorum**.

Initially, C, D, and E all store version **v1** of k. Every successful write creates a new version. We label these versions in the order an outside observer sees them written; these labels do not necessarily tell us how the system determines which version is newer.

A network failure now splits the cluster into two groups:

* **A, B, and C** can communicate with one another.
* **D and E** can communicate with each other.
* No communication is possible between the two groups. Each group treats the nodes in the other group as unreachable.

**a.** A client in the A–B–C group writes a new version, **v2**. Which node coordinates the write? Which nodes are contacted, and which nodes store v2?

**b.** At the same time, a client in the D–E group reads k. Which node coordinates the read? Which nodes respond, and which version does the client receive?

**c.** Although **R + W = 4 > N = 3**, the second client does not see v2. Why does this happen? What does a sloppy quorum still guarantee, and what does it fail to guarantee?

**d.** The client in the D–E group now writes **v3**, based on the version it just read. Where is v3 stored?

Once communication between the groups is restored, C coordinates a new read of k. Which versions does it receive? Can the system consider one version newer than the other? What happens to the copies of k stored on A and B?

**e.** During the network partition, which CAP property does the system give up: **consistency (C), availability (A), or partition tolerance (P)**?

**f.** Now suppose the system uses a **strict quorum**: unreachable nodes are never replaced, and only C, D, and E may be contacted for k.

During the partition, what happens to each of these requests?

* The write of v2 in part a.
* The read in part b.
* The write of v3 in part d.

Which CAP property does the system give up in this case?


#### Solution

**a.** C coordinates the write because it is the first reachable node in the preference list.

C cannot reach D or E, so it uses the next available nodes around the ring: A and B. C stores v2 locally and sends it to both.

The write succeeds once two nodes have confirmed that they stored v2. For example, C’s own copy plus A’s confirmation satisfy **W = 2**. B also stores v2 when it processes the write.

A and B are temporarily storing a key that would normally belong on C, D, and E.

**b.** D coordinates the read because it is the first reachable node in the preference list on this side of the partition.

D reads its own copy and asks E for its copy. Both hold v1. Their two responses satisfy **R = 2**, so the client receives **v1**.

This version is out of date, but D and E have no way to know that v2 exists.

**c.** The rule **R + W > N** guarantees overlap only when reads and writes use the same fixed set of N nodes.

Here, the write can succeed using **C and A**, while the read succeeds using **D and E**. These sets have no nodes in common, so the read does not discover v2.

A sloppy quorum allows both groups to keep serving requests in this example. Each successful write is confirmed by at least W nodes, but those nodes may be temporary replacements.

It does **not** guarantee that a read returns the latest successful write. The system provides eventual consistency: once communication is restored and conflicts are resolved, the copies can converge.

**d.** D coordinates the new write and stores **v3 on D and E**.

After the partition ends, C coordinates a read. It finds:

* **v2** in its own storage.
* **v3** on D or E.

Neither version is a later revision of the other. Both were created from v1, without knowing about the other update. They are therefore **concurrent versions**.

The names v2 and v3 reflect the order in which an outside observer saw the writes. They do not mean that v3 includes or replaces the changes in v2.

Dynamo uses **vector clocks** to distinguish between:

* An older version that has been replaced by a later revision.
* Independent versions that contain conflicting changes.

A simple version counter cannot reliably make this distinction. A larger counter on one side does not mean that its version includes the other side’s changes.

C returns both versions to the client. The client resolves the conflict and writes the combined result as **v4**.

Meanwhile, A and B transfer their temporary copies to the intended storage nodes, D and E. Once the transfer succeeds, A and B delete those temporary copies. This process is called **hinted handoff**.

**e.** The system gives up **consistency (C)** during the partition.

Both groups continue to serve reads and writes, but a read may miss a successful write made in the other group. In CAP terms, this is **AP** behavior: the system preserves availability while tolerating the partition.

**f.** With a strict quorum, only **C, D, and E** can store or serve k. A and B cannot act as replacements.

The three requests behave as follows:

| Request                       | Reachable storage nodes | Result                                      |
| ----------------------------- | ----------------------- | ------------------------------------------- |
| Write v2 from the A–B–C group | C only                  | Cannot succeed: it needs two confirmations. |
| Read from the D–E group       | D and E                 | Succeeds and returns v1.                    |
| Write v3 from the D–E group   | D and E                 | Succeeds: both nodes confirm the write.     |

The read returns v1 because the attempted write of v2 never succeeded. In this sequence, the successful writes do not create conflicting branches.

The A–B–C group cannot successfully read or write k during the partition because it can reach only one of the three storage nodes.

The system therefore gives up **availability (A)** on that side. Under the exercise’s assumptions, this is **CP** behavior: it preserves consistency while tolerating the partition.

Strict quorum overlap alone is not a complete guarantee of CAP consistency in every implementation; that also depends on how the system handles versions, concurrent operations, and incomplete writes.
